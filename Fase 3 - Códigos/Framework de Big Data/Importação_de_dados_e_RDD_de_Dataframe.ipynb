# Criando a sessão do SparkContext e SparkSession

```python
from pyspark import SparkContext
from pyspark.sql import SparkSession
```

```python
sc = SparkSession.builder.appName('pySpark DataFrame From RDD').getOrCreate()
```

# Create PySpark DataFrame from an Existing RDD

```python
rdd = sc.sparkContext.parallelize([('C', 85,76,87,91),('B', 85,76,87,91),('A', 85,76,87,91)], 4)
print(type(rdd))
```

```python
sub = ['id_person','value_1','value_2','value_3', 'value_4']
```

```python
marks_df = sc.createDataFrame(rdd, schema=sub)
```

```python
print(type(marks_df))
```

```python
marks_df.printSchema()
```

```python
marks_df.show()
```

# Creating and Manipulation in PySpark DataFrame

```python
!pip install pyspark
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('pysparkdf').getOrCreate()
```

# Import Data

```python
df = spark.read.csv('/content/cereal.csv', header=True, inferSchema=True)
```

# Select()

```python
df.select('name', 'mfr', 'rating').show()
```

# withColumn()

```python
df.withColumn('Calories', df['calories'].cast('Integer')).printSchema()
```

# groupBy()

```python
df.groupBy('calories').count().show()
```

# orderBy()

```python
df.orderBy('protein').show()
```

# Case When

```python
from pyspark.sql.functions import when
```

```python
df.select('name', df.vitamins, when(df.vitamins >="25", "rich in vitamins")).show()
```

# filter()

```python
df.filter(df.calories >= "100").show()
```

# isnull() / isnotnull()

```python
from pyspark.sql.functions import *
```

```python
df.filter(df.name.isNull()).show()
```

```python
df.filter(df.name.isNotNull()).show()
```
