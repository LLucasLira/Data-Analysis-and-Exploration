## 2. Proporcionalidade e Seaborn

```python
dados_usaveis = pd.read_csv(r"e:\PÓS GRADUAÇÃO\Fase 1 - Data Analysis and Exploration\3. Visualização de dados\Dados\sih_cnv_qiuf135059177_102_198_52.csv", sep = ";", skiprows= 3,encoding = "iSO-8859-1")
dados_usaveis.head()
```

```python
ordenados_por_total = dados_usaveis.sort_values(by="Total", ascending=False)
ordenados_por_total = ordenados_por_total.drop("Total", axis=1)
ordenados_por_total = ordenados_por_total.dropna()
ordenados_por_total.head(5)
```

```python
colunas_interessadas = ordenados_por_total.columns
ordenados_por_total = ordenados_por_total[colunas_interessadas]
colunas_interessadas
```

```python
# Separar a primeira coluna (preservar como está)
primeira_coluna = ordenados_por_total.iloc[:, 0]

# Aplicar replace e conversão apenas nas colunas numéricas (do 2º em diante)
colunas_numericas = ordenados_por_total.iloc[:, 1:]
colunas_numericas = colunas_numericas.replace(",", ".", regex=True).apply(pd.to_numeric, errors='coerce') / 1_000_000

# Reunir tudo novamente
ordenados_por_total = pd.concat([primeira_coluna, colunas_numericas], axis=1)

# Exibir o resultado
ordenados_por_total.head()
```

```python
coluna_nome = ordenados_por_total.columns[0]
ordenados_por_total[coluna_nome] = ordenados_por_total[coluna_nome].astype(str).str.replace(r'\d+', '', regex=True).str.strip()
```

```python
ordenados_por_total = ordenados_por_total.set_index('Unidade da Federação')
ordenados_por_total.head()
```

```python
mes_mais_recente = ordenados_por_total.columns[-1]
mes_mais_recente
```

```python
gastos_do_mais_recente = ordenados_por_total[mes_mais_recente]
gastos_do_mais_recente.head()
```

```python
for estado in gastos_do_mais_recente.index:
  dados_da_populacao_io.index = dados_da_populacao_io.index.str.replace(f"{estado} {estado}", estado)
dados_da_populacao_io.index
```

```python
gastos_e_populacao = dados_da_populacao_io.join(gastos_do_mais_recente)
gastos_e_populacao.head()
```

```python
ultima_coluna = gastos_e_populacao.columns[-1]
gastos_e_populacao["gastos"] = gastos_e_populacao[ultima_coluna] * 1_000_000
gastos_e_populacao.plot()
```

```python
gastos_e_populacao["gasto_por_habitante"] = gastos_e_populacao["gastos"] / gastos_e_populacao["populacao"]
gastos_e_populacao.head()
```

## 4. Trabalhando Datetime e Melt

```python
mensal.info()
```

```python
mensal.index
```

```python
from datetime import date

meses = {
    "Jan" : 1,
    "Fev" : 2,
    "Mar" : 3,
    "Abr" : 4,
    "Mai" : 5,
    "Jun" : 6,
    "Jul" : 7,
    "Ago" : 8,
    "Set" : 9,
    "Out" : 10,
    "Nov" : 11,
    "Dez" : 12
}

def para_dia(ano_mes: str):
  ano: int = int(ano_mes[:4])
  mes_como_string: str = ano_mes[5:]
  mes: int = meses[mes_como_string]
  return date(ano, mes, 1)
```

```python
mensal.index.map(para_dia)
```

```python
mensal.index = mensal.index.map(para_dia)
mensal.head()
```

```python
plt.figure(figsize=(10,6))
axis = sns.lineplot(data=mensal, x=mensal.index, y="São Paulo")
plt.ylim(0, 600)
plt.grid(linestyle="--")
plt.show()
```

```python
mensal.index[0]
```

```python
plt.figure(figsize=(10,6))
axis = sns.lineplot(data=mensal, x=mensal.index, y="São Paulo")
axis = sns.lineplot(data=mensal, x=mensal.index, y="Minas Gerais")
plt.ylim(0, 600)
plt.grid(linestyle="--")
plt.show()
```

```python
mensal.head()
```

```python
mensal_aberto = mensal.reset_index().melt(id_vars=["index"], value_vars=mensal.columns)
mensal_aberto.columns = ["dia_mes_ano", "uf", "gasto"]
mensal_aberto.head()
```

```python
plt.figure(figsize=(10,6))
axis = sns.lineplot(data=mensal_aberto, x="dia_mes_ano", y="gasto", hue="uf")
plt.ylim(0, 600)
plt.grid(linestyle="--")
plt.show()
```
